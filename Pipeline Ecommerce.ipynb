```python
#===========================================
# Pipeline e-commerce
#===========================================

#Liste brutes des montants de commandes reçues aujurd'hui en fcfa
commandes=[5000,10000,3000,2500,4000,7000,3400,2300,200000,1000]
print("=== RAPPORT DES COMMANDES ===")
print("Nombre de commandes :",len(commandes))
print("Première commande  : ", commandes [0] ,"fcfa")
print("Dernière commande  :", commandes [-1] ,"fcfa")
```

```python
# ============================================
# SLICING — Découper la liste des commandes
# ============================================

# Les 3 premières commandes
print("3 premières commandes :", commandes[0:3])

# Les 3 dernières commandes
print("3 dernières commandes :", commandes[-3:])

# Une commande sur deux
print("Une commande sur deux :", commandes[::2])

# La liste à l'envers
print("Commandes inversées :", commandes[::-1])
```

```python
# ============================================
# MODIFICATION — Ajouter & Supprimer
# ============================================
commandes=[5000,10000,3000,2500,4000,7000,3400,2300,200000,1000]
print("Avant :", commandes)

# Ajouter une nouvelle commande à la fin
commandes.append(15000)
print("Après append :", commandes)

# Supprimer la dernière commande
commandes.pop()
print("Après pop :", commandes)

# Insérer une commande à la position 2
commandes.insert(2, 8000)
print("Après insert :", commandes)
```

```python
# ============================================
# BILAN — Analyse des commandes
# ============================================
commandes=[5000,10000,3000,2500,4000,7000,3400,2300,200000,1000]
print("=== BILAN DU JOUR ===")
print("Nombre de commandes :", len(commandes))
print("Commande minimum :", min(commandes), "fcfa")
print("Commande maximum :", max(commandes), "fcfa")
print("Total des ventes :", sum(commandes), "fcfa")
```

```python
# ============================================
# MODULE 2 — Fiches clients (Dictionnaires)
# ============================================

# Fiche d'un client
client = {
    "nom" : "Diallo",
    "prenom" : "Mamadou",
    "ville" : "Cotonou",
    "montant": 5000
}

print("=== FICHE CLIENT ===")
print("Nom :", client["nom"])
print("Ville :", client["ville"])
print("Montant:", client["montant"], "fcfa")
```

```python
# ============================================
# MODIFICATION — Enrichir la fiche client
# ============================================

# Ajouter un nouveau champ
client["telephone"] = "0022997000000"
print("Après ajout téléphone :", client)

# Modifier une valeur existante
client["montant"] = 7500
print("Après modification montant :", client["montant"], "fcfa")

# Lire une clé qui n'existe pas — méthode sécurisée
email = client.get("email", "email non renseigné")
print("Email :", email)
```

```python
# ============================================
# DICTIONNAIRES IMBRIQUÉS — Plusieurs clients
# ============================================

# Base de données de 3 clients
clients = {
    "C001": {
        "nom" : "Diallo",
        "ville" : "Cotonou",
        "montant": 5000
    },
    "C002": {
        "nom" : "Charle",
        "ville" : "Bohicon",
        "montant": 12000
    },
    "C003": {
        "nom" : "Habib",
        "ville" : "Parakou",
        "montant": 8500
    }
}

print("=== BASE CLIENTS ===")
print("Client C001 :", clients["C001"]["nom"], "-", clients["C001"]["montant"], "fcfa")
print("Client C002 :", clients["C002"]["nom"], "-", clients["C002"]["montant"], "fcfa")
print("Client C003 :", clients["C003"]["nom"], "-", clients["C003"]["montant"], "fcfa")
```

```python
# ============================================
# MODULE 3 — Filtrage des commandes (Boucles)
# ============================================

# Liste de toutes les commandes avec détails
commandes_details = [
    {"id": "C001", "nom": "Diallo", "montant": 5000, "statut": "payé"},
    {"id": "C002", "nom": "Charle", "montant": 12000, "statut": "en attente"},
    {"id": "C003", "nom": "Habib", "montant": 8500, "statut": "payé"},
    {"id": "C004", "nom": "Koné", "montant": 2000, "statut": "annulé"},
    {"id": "C005", "nom": "Bénoit", "montant": 15000, "statut": "payé"},
    {"id": "C006", "nom": "Franck", "montant": 3500, "statut": "en attente"},
    {"id": "C007", "nom": "Sow", "montant": 25000, "statut": "payé"},
    {"id": "C008", "nom": "Bello", "montant": 1500, "statut": "annulé"},
]

# Filtrer uniquement les commandes payées ET supérieures à 5000 fcfa
print("=== COMMANDES PAYÉES > 5000 FCFA ===")
for commande in commandes_details:
    if commande["statut"] == "payé" and commande["montant"] > 5000:
        print(commande["nom"], "-", commande["montant"], "fcfa")
```

```python
# ============================================
# COMPRÉHENSIONS — Version courte et élégante
# ============================================

# Version longue (boucle classique)
gros_clients = []
for commande in commandes_details:
    if commande["montant"] > 5000:
        gros_clients.append(commande["nom"])

print("Version longue :", gros_clients)

# Version courte (compréhension) — même résultat en 1 ligne !
gros_clients2 = [c["nom"] for c in commandes_details if c["montant"] > 5000]

print("Version courte :", gros_clients2)
```

```python
# ============================================
# MODULE 4 — Nettoyage (Sets, Tuples, Tri)
# ============================================

# Liste avec des doublons (clients qui ont commandé plusieurs fois)
ids_clients = ["C001", "C003", "C001", "C002", "C003", "C001", "C004"]

print("Avant nettoyage :", ids_clients)

# Supprimer les doublons avec un SET
ids_uniques = set(ids_clients)
print("Sans doublons :", ids_uniques)

# TUPLE — informations fixes d'un produit (on ne peut pas modifier)
produit = ("Chaussures Nike", "REF-2024", 35000)
print("Produit :", produit[0], "- Réf:", produit[1], "- Prix:", produit[2], "fcfa")

# TRI — trier les commandes par montant croissant
commandes_triees = sorted(commandes_details, key=lambda c: c["montant"])
print("\n=== COMMANDES TRIÉES PAR MONTANT ===")
for c in commandes_triees:
    print(c["nom"], "-", c["montant"], "fcfa")
```
